```python
import numpy as np
import pandas as pd
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, SimpleRNN, LSTM, GRU, Dense
from tensorflow.keras.preprocessing.sequence import pad_sequences

# 1. Load IMDB dataset
vocab_size = 10000
max_length = 200

(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=vocab_size)

X_train = pad_sequences(X_train, maxlen=max_length)
X_test = pad_sequences(X_test, maxlen=max_length)

# 2. Define models
layers = {
    'Simple RNN': SimpleRNN(32),
    'LSTM': LSTM(32),
    'GRU': GRU(32)
}

results = []

# 3. Train and evaluate models
for name, layer in layers.items():

    model = Sequential([
        Input(shape=(max_length,)),
        Embedding(vocab_size, 32),
        layer,
        Dense(1, activation='sigmoid')
    ])

    model.compile(
        optimizer='adam',
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    print("\nTraining", name)

    model.fit(
        X_train, y_train,
        epochs=5,
        batch_size=128,
        validation_split=0.2,
        verbose=1
    )

    loss, accuracy = model.evaluate(X_test, y_test, verbose=0)

    results.append([
        name,
        round(accuracy * 100, 2),
        round(loss, 4)
    ])

# 4. Performance comparison table
df = pd.DataFrame(results, columns=[
    'Model', 'Test Accuracy (%)', 'Test Loss'
])

print("\nPERFORMANCE COMPARISON")
print(df.to_string(index=False))

# 5. Final analysis table
best_acc = df.loc[df['Test Accuracy (%)'].idxmax()]
best_loss = df.loc[df['Test Loss'].idxmin()]

final_analysis = pd.DataFrame({
    'Performance': ['Best Test Accuracy', 'Lowest Test Loss'],
    'Best Model': [best_acc['Model'], best_loss['Model']]
})

print("\nFINAL ANALYSIS")
print(final_analysis.to_string(index=False))
```
